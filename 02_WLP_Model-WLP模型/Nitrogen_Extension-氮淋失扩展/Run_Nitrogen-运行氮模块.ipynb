# Run Nitrogen-运行氮模块

Default: strict Aachen inputs, with unknown values blocking execution. Explicitly enable the published reference only for technical checking; its results do not describe the Aachen field. / 默认严格地块输入，缺项会阻止运行；仅技术检查时显式启用官方示例，不代表Aachen地块。


```python
from pathlib import Path
import runpy
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "06_Hao_Interface_and_Rerun-Hao接口与复跑").is_dir())
registry = runpy.run_path(str(root / "06_Hao_Interface_and_Rerun-Hao接口与复跑/Shared_Core-共享核心/Module_Loader-模块加载.py"))
registry["install"]()
n = registry["load_module"]("nitrogen_engine")
USE_PUBLISHED_REFERENCE = False
crop_ids = list(n.base.crop_catalog())
```

```python
results = []
for crop_id in crop_ids:
    config = n.NitrogenConfig(crop_id=crop_id,
        nitrogen_inputs=n.REFERENCE if USE_PUBLISHED_REFERENCE else n.DEFAULT_FIELD,
        accept_reference_scenario=USE_PUBLISHED_REFERENCE, accept_demo_assumptions=True)
    issues = n.preflight(n.load_inputs(config), config)
    if issues:
        results.append({"crop_id": crop_id, "status": "blocked_missing_sourced_inputs", "missing": issues})
    else:
        payload, folder = n.evaluate(config)
        results.append({"crop_id": crop_id, "status": payload["status"],
            "leaching_kg_N_ha": payload["nitrogen_leaching_kg_N_ha"], "folder": folder.relative_to(root).as_posix()})
results
```
